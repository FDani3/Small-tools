```python
# Get the soup that i pasted from my browser and saved
from bs4 import BeautifulSoup

html_doc = open("/content/temp.html", "r")

soup = BeautifulSoup(html_doc, 'html.parser')

#make a list of the tag of the journals
journal_tags = soup.find_all('a', {"class" : "text-primary-400 text-sm hover:text-primary-200"})
journals = []
for i in journal_tags:
  journals.append(i)
```

```python
# Get the pandas of the xlsx of the cs journals
import pandas as pd

# Load the XLSX file
cs_journals = pd.read_excel('/content/1472JournalsOfComputerfromallpublishers.xlsx')

# Display the first few rows
print(str(cs_journals["Journals Name"][0]))
```

```python
# Get matches
# It is a bit sloppy rn
import re

def match_journals(jc, j):
  if jc in j:
    return True
  if j in jc:
    return True
  # Sometimes you find "Advanced Academics, Journal of" type of things
  # you want these rearranged
  j = re.sub(r"^(.*),\s*(.*)$", r"\2 \1", j)
  if jc in j:
    return True
  if j in jc:
    return True

  return False


# Check if there is a match, save the result, brute force
results = dict()

for j in journals:
  for jc in cs_journals["Journals Name"]:
    if match_journals(jc, j.contents[0]):
      results[jc] = True
```

```python
left_out = [x[1]["Journals Name"] for x in cs_journals.iterrows() if x[1]["Journals Name"] not in results and x[1]["Open Access status"] != "Unpaywall Open Acess"]
```

```python
print("How many you cant access: ", len(left_out), "How many matches you have: " len(results))
```
